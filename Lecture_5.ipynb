## 5.1 Intro

A function is a block of code which only runs when it is called.

You can pass data, known as parameters, into a function.

A function can return data as a result.

```python
def <function name>(<arguments>):
    <function body>
    return <values> # can be omitted. Depend on task
```

````python
def print_name (name: str)-> None: # before Python 3.11 def print_name (name)
    print(f"Name is {name}")
````

````python
print_name("Python")
````

````python
print_name(45)
````

````python
def print_name_comment (name: str)-> None:
    """
    A function which prints the value of the variable `str` like 'Name is `str`'
    """
    print(f"Name is {name}")
````

````python
print_name_comment("Python")
````

````python
def print_value(value: int | float) -> None:
    print(value)
````

The function definitions cannot be empty, but if you for some reason have a function definition with no content, put in the `pass` statement to avoid getting an error.

```python
def <function name>(<argument>):
    pass
```

````python
def pass_test():
    pass
````

````python
pass_test()
````

## 5.2 Positional and Keyword Arguments

To specify that a function can have only positional arguments, add `, /` after the arguments:

```python
def <function name>(<positional arguments>, /):
    <function body>
```

````python
def print_value_mag(value: int | float, mag: int | float, /) -> None:
    print(value * mag)
````

To specify that a function can have only keyword arguments, add `*,` before the arguments:

```python
def <function name>(*, <keyword arguments>):
    <function body>
```

````python
def print_value_mag(*, value: int | float, mag: int | float = 5) -> None:
    print(value * mag)
````

Any argument before the `/,` are positional-only, and any argument after the `*,` are keyword-only:

```python
def <function name>(<positional arguments>, /, *, <keyword arguments>):
    <function body>
```

````python
def print_value_mag( value: int | float,  mag: int | float=5) -> None:
    print(value * mag)
````

````python
print_value_mag(10, mag=10)
````

````python
print_value_mag(mag=12, value=10 )
````

## 5.3 Arbitrary Arguments (*args) and Arbitrary Keyword Arguments (**kwargs)

If you do not know how many arguments that will be passed into your function, add a * before the parameter name in the function definition.

This way the function will receive a tuple of arguments, and can access the items accordingly:

```python
def <function name>(*args, **kwargs):
    <function body>
```

````python
def print_values(*args : int | float)->float|int:
    s=0
    for i in args:
        s+=i
    return s
````

````python
def f(x,y):
    return x**2+y**2
````

````python
x=2
y = 3
z = f(x,y)
````

````python
print(z)
````

````python
a = print_values(1,5)
````

````python
print_values(1, 2, 5, 10.1, "NANME" )
````

If you do not know how many keyword arguments that will be passed into your function, add two asterisk: ** before the parameter name in the function definition.

This way the function will receive a dictionary of arguments, and can access the items accordingly:


```python
def <function name>(**kwargs):
    <function body>
```

````python
def print_values_kwarg(**kwargs : int | float):
    print(kwargs["value"] * kwargs["mag"])
````

````python
print_values_kwarg(value=5, mag=-5)
````

## 5.6 Recursion

Recursion is a common mathematical and programming concept. It means that a function calls itself. This has the benefit of meaning that you can loop through data to reach a result.

```python
def <function name>(<argument>):
    <function body>
    return <function name>(<argument>) # can be without return
```

````python
def fact(n):
    if n in (0, 1):
        return 1
    return fact(n - 1) * n
````

````python
val = fact(10)
````

### Recursion depth

````python
def fact_depth(n):
    if n in (0, 1):
        return 1
    return fact_depth(n - 1) * n % 1e9
````

````python
val = fact_depth(10000)
# val = fact_depth(100000)
````

````python
from sys import setrecursionlimit
````

````python
setrecursionlimit(20000)
val = fact_depth(10000)
````

## 5.7 Cache

````python
%%timeit
val = fact(500)
````

````python
%%timeit
val = fact(500)
````

````python
from functools import lru_cache

@lru_cache(maxsize=1000)
def fact(n):
    if n in (0, 1):
        return 1
    return fact(n - 1) * n
````

````python
%%timeit
val = fact(500)
````

````python
%%timeit
val = fact(500)
````

````python
%%timeit
val = fact(600)
````

## 5.8 Lambda function

A lambda function is a small anonymous function.

A lambda function can take any number of arguments, but can only have one expression.

```python
<name> = lambda <arguments> : <expression>
```

````python
y = lambda x: x**2
````

````python
def y(x: int)-> int:
    return x**2
````

````python
y(5)
````

````python
x = list(range(-10, 11, 1))
print(*list(map(y, x)))
````

````python
y = lambda x: x**2 + 2 * x + 3
````

````python
def y(x: int) -> int:
    return x**2 + 2 * x + 3
````

````python
x = list(range(-10, 11, 1))
print(*list(map(y, x)))
````

````python
y = lambda x, a, b, c: a * x**2 + b * x + c
````

````python
def y(x: int, a: int, b: int, c: int)->int:
    return a * x**2 + b * x + c
````

````python
y(5, 1, 2, 3)
````

````python
print(*list(map(y, x, [1] * len(x), [2] * len(x), [3] * len(x))))
````
